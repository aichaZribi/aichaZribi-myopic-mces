```python
from pathlib import Path
import re

import pandas as pd
import matplotlib.pyplot as plt


# --------------------------------------------------
# File locations
# --------------------------------------------------
DATA_DIR = Path("../../example/filter_mass_diff_and_size")

FILES = [
    DATA_DIR / "filtered_pairs_200_300_CPLEX_PY_threads_8_original.csv",
    DATA_DIR / "filtered_pairs_300_500_HiGHS_CMD_threads_1_original.csv",
    DATA_DIR / "filtered_pairs_500_800_HiGHS_CMD_threads_1_original.csv",
    DATA_DIR / "filtered_pairs_800_1000_HiGHS_CMD_threads_1_originalchecl.csv",
]


# --------------------------------------------------
# Load files and extract mass range from filename
# --------------------------------------------------
dataframes = []

for file_path in FILES:

    if not file_path.exists():
        raise FileNotFoundError(
            f"File not found: {file_path.resolve()}"
        )

    df = pd.read_csv(file_path)

    # Extract values such as 200 and 300 from the filename
    match = re.search(
        r"filtered_pairs_(\d+)_(\d+)_",
        file_path.name
    )

    if match is None:
        raise ValueError(
            f"Could not extract mass range from: {file_path.name}"
        )

    lower_mass = int(match.group(1))
    upper_mass = int(match.group(2))

    df["mass_range"] = f"{lower_mass}-{upper_mass}"
    df["lower_mass"] = lower_mass
    df["upper_mass"] = upper_mass
    df["range_midpoint"] = (lower_mass + upper_mass) / 2
    df["source_file"] = file_path.name

    dataframes.append(df)


all_sizes = pd.concat(
    dataframes,
    ignore_index=True
)

display(all_sizes.head())
```

```python
range_order = [
    "200-300",
    "300-500",
    "500-800",
    "800-1000",
]

all_sizes["mass_range"] = pd.Categorical(
    all_sizes["mass_range"],
    categories=range_order,
    ordered=True
)
```

```python
runtime_summary = (
    all_sizes
    .groupby(
        "mass_range",
        observed=True
    )
    .agg(
        number_of_pairs=("pair_id", "count"),
        mean_runtime_sec=("runtime_sec", "mean"),
        median_runtime_sec=("runtime_sec", "median"),
        minimum_runtime_sec=("runtime_sec", "min"),
        maximum_runtime_sec=("runtime_sec", "max"),
        timeout_count=("skipped", "sum"),
        solver_calls=("solverCalled", "sum"),
    )
    .reset_index()
)

runtime_summary["timeout_rate_percent"] = (
    100
    * runtime_summary["timeout_count"]
    / runtime_summary["number_of_pairs"]
)

runtime_summary = runtime_summary.round(3)

display(runtime_summary)
```

```python
plot_data = [
    all_sizes.loc[
        all_sizes["mass_range"] == mass_range,
        "runtime_sec"
    ].dropna()
    for mass_range in range_order
]

plt.figure(figsize=(9, 6))

plt.boxplot(
    plot_data,
    tick_labels=range_order,
    showfliers=True
)

plt.yscale("log")

plt.xlabel("Molecular mass range")
plt.ylabel("Runtime in seconds — logarithmic scale")
plt.title("Runtime Distribution by Molecular Mass Range")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```
